### **`Encapsulation`**

(Boundry and self - sovereignity) It defines the membrane of identity around an object which is a sovereign entity where state and behaviour forms an indespensible unit, without encapsulation an object's boundry dissolve into global state and it cease to exist as an coherent, self governing thing. 

(Invariant protection and change isolation) It hides internal structure (protected and private fields) and forces interactions though validated methods, resulting in predictable behaviour.


E.g. : A car engine has a lot of internals like fuel injection that is manages itself and not by the user of the car.

```python
# Bad Example, balance attribute is open and can be manipulated anyways, introducing chaoctic behaviour

class BadBankAccount:
  def __init__(self, balance):
    self.balance = balance

account = BadBankAccount(0.0)
account.balance = -1 # Oh dear -- balance should not be allowed to be negative
print(account.balance)
```

```python
# Properly encapsulated Class

class BankAccount:

    MINIMUN_BALANCE = 10
    
    def __init__(self, balance, type):
        self._balance = balance # Should be declared public to 
        self.type = type

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError(f"Amount : {amount} must be more then zero")
        else:
            self._balance += amount

    def withdraw(self, amount):
        if amount > 0:            
            if self._balance - amount < BankAccount.MINIMUN_BALANCE:
                raise ValueError(f"With amount : {amount} , new balance is less than minimum balance : {BankAccount.MINIMUN_BALANCE} ")
            else:
                self._balance -= amount
        else:
            raise ValueError(f"Amount : {amount} must be more then zero")



new_account = BankAccount(10, "Savings")
new_account.deposit(10)
print(f"New Balance : {new_account.balance}")
new_account.deposit(5)
print(f"New Balance : {new_account.balance}")
new_account.withdraw(10)
print(f"New Balance : {new_account.balance}")
```

### **`Abstraction`**

(Essence vs. Accident) It separates an entity's essential nature from its accidental particulars. It defines what an entity fundamentally is by stripping away irrelevant mechanical details.

(Perspective & Cognitive Load): Abstraction is strictly relative to the observer's context. Via interfaces and abstract contracts keeps focus on high-level domain only and hides implementation details.

```python
# Bad example, all methods are exposed and this complicates things

class BadEmailService:

    def __int__(self):
        pass

    def connect(self):
        print("Connecting to email server...")

    def authenticate(self):
        print("Authenticating...")

    def send_email(self):
        print("Sending email...")

    def disconnect(self):
        print("Disconnecting from email server...")

email = BadEmailService()

email.connect()
email.authenticate()
email.send_email()
email.disconnect()
```

```python
# Properly abstracted class, abstracting the email sending process with hidden implementtion details

class EmailService:

    def __int__(self):
        pass

    def _connect(self):
        print("Connecting to email server...")

    def _authenticate(self):
        print("Authenticating...")

    def _disconnect(self):
        print("Disconnecting from email server...")

    def send_email(self):
        self._connect()
        self._authenticate()
        print("Sending email...")
        self._disconnect()

email = EmailService()
email.send_email()
```

### **`Polymorphism`**

(Multiplicity of Form under Unity of Meaning) Allows a single universal concept or action to manifest through multiple distinct concrete realities Each entity fulfills the same philosophical contract according to its own nature.

(Decoupled Execution & Extensibility): Contextually, polymorphism eliminates brittle if/else or switch chains that inspect object types. A checkout service simply calls paymentMethod.process(amount) without knowing whether the runtime object is a credit card, cryptocurrency wallet, or bank wire. This directly enables the Open/Closed Principle: you can introduce brand-new behaviors to a system without modifying the existing code that orchestrates them.

```python
# Bad example without polymorphism

class Car:
    def __init__(self, brand, model, year, number_of_doors):
        self.brand = brand
        self.model = model
        self.year = year
        self.number_of_doors = number_of_doors

    def start(self):
        print("Car is starting.")

    def stop(self):
        print("Car is stopping.")

class Motorcycle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start_bike(self):
        print("Motorcycle is starting.")

    def stop_bike(self):
        print("Motorcycle is stopping.")

# Create list of vehicles to inspect
vehicles = [
    Car("Ford", "Focus", 2008, 5),
    Motorcycle("Honda", "Scoopy", 2018)
]

# Loop through list of vehicles and inspect them
for vehicle in vehicles:
    if isinstance(vehicle, Car):
        print(f"Inspecting {vehicle.brand} {vehicle.model} ({type(vehicle).__name__})")
        vehicle.start()
        vehicle.stop()
    elif isinstance(vehicle, Motorcycle):
        print(f"Inspecting {vehicle.brand} {vehicle.model} ({type(vehicle).__name__})")
        vehicle.start_bike()
        vehicle.stop_bike()
    else:
        raise Exception("Object is not a valid vehicle")
    
```

```python
# Good example using polymorphism

# Parent class ("Superclass")
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start(self):
        print("Vehicle is starting.")

    def stop(self):
        print("Vehicle is stopping.")

# Child class ("Subclass") of Vehicle superclass
class Car(Vehicle):
    def __init__(self, brand, model, year, number_of_doors):
        super().__init__(brand, model, year) # Initialising parent class's attributes.
        self.number_of_doors = number_of_doors

    # Below, we "override" the start and stop methods, inherited from Vehicle, to provide car-specific behaviour

    def start(self):
        print("Car is starting.")

    def stop(self):
        print("Car is stopping.")

# Child class ("Subclass") of Vehicle superclass
class Motorcycle(Vehicle):
    def __init__(self, brand, model, year):
        super().__init__(brand, model, year)

    # Below, we "override" the start and stop methods, inherited from Vehicle, to provide bike-specific behaviour

    def start(self):
        print("Motorcycle is starting.")

    def stop(self):
        print("Motorcycle is stopping.")

vehicles = [
    Car("Ford", "Focus", 2008, 5),
    Motorcycle("Honda", "Scoopy", 2018)
]

# Loop through list of vehicles and inspect them
for vehicle in vehicles:
    if isinstance(vehicle, Vehicle):
        print(f"Inspecting {vehicle.brand} {vehicle.model} ({type(vehicle).__name__})")
        vehicle.start()
        vehicle.stop()
    else:
        raise Exception("Object is not a valid vehicle")
```

### **`Inheritance`**

(Taxonomy & Lineage — The "Is-A" Relation): Inheritance establishes a vertical hierarchy of being, mirroring biological taxonomy.

(Subtyping vs. The Fragile Base Class Trap): Contextually, inheritance enables code reuse. However, business domains rarely obey neat, permanent taxonomies. When requirements shift, deep inheritance trees become brittle—a minor modification to a parent class can unexpectedly break dozens of subclasses (the Fragile Base Class problem).

### **`Composition`**


("Has-A" Relation) It defines an entity not by its ancestral bloodline, but by the constituent parts assembled within it. A Drone does not inherit from GPSModule or Battery; ontologically, a Drone is a whole composed of a flight controller, propulsion system, and sensor payload.

(Modular Assembly & Runtime Flexibility) In modern software engineering, the maxim "favor composition over inheritance" is a pragmatic survival strategy. 


### **`Coupling`**

(Existential Entanglement): Coupling measures relational independence—the degree to which Entity A's existence, structure, or definition depends on knowledge of Entity B. Under tight coupling, two entities are ontologically entangled; neither can be understood or instantiated in isolation. Under loose coupling, entities interact solely through abstract boundaries, preserving independent identity.

(Blast Radius & Cost of Change): In day-to-day engineering, coupling determines the blast radius of a code change. If an OrderService directly instantiates a MySQLConnection and writes raw SQL queries inside its business logic, changing databases requires rewriting the business logic
