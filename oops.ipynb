#### Encapsulation

The concept of boundry and self determination, an object has an indide that belongs to it. It is data protection and stage management bundled together. A class is supposed to be interacted with simple methods rather than reaching for their internal details giving predictability

E.g. : A car engine has a lot of internals like fuel injection that is manages itself and not by the user of the car.

```python
# Bad Example, balance attribute is open and can be manipulated anyways, introducing chaoctic behaviour

class BadBankAccount:
  def __init__(self, balance):
    self.balance = balance

account = BadBankAccount(0.0)
account.balance = -1 # Oh dear -- balance should not be allowed to be negative
print(account.balance)

```

```python
# Properly encapsulated Class

class BankAccount:

    MINIMUN_BALANCE = 10
    
    def __init__(self, balance, type):
        self._balance = balance # Should be declared public to 
        self.type = type

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError(f"Amount : {amount} must be more then zero")
        else:
            self._balance += amount

    def withdraw(self, amount):
        if amount > 0:            
            if self._balance - amount < BankAccount.MINIMUN_BALANCE:
                raise ValueError(f"With amount : {amount} , new balance is less than minimum balance : {BankAccount.MINIMUN_BALANCE} ")
            else:
                self._balance -= amount
        else:
            raise ValueError(f"Amount : {amount} must be more then zero")



new_account = BankAccount(10, "Savings")
new_account.deposit(10)
print(f"New Balance : {new_account.balance}")
new_account.deposit(5)
print(f"New Balance : {new_account.balance}")
new_account.withdraw(10)
print(f"New Balance : {new_account.balance}")
```
