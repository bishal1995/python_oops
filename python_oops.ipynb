Classes are blueprints for objects, they have attributes (informations) and methods(behaviours) 

`__init__(self)` : method that runs only once when the object is instantiated. it can take parameters

self is a special parameter that refers to the instance of the class (the object) you're working with. When you define a method within a class, the first parameter of that method is always self, by convention. This helps Python know that the method belongs to an instance of the class.
