`__init__(self)` : method that runs only once when the object is instantiated. it can take parameters

self is a special parameter that refers to the instance of the class (the object) you're working with. When you define a method within a class, the first parameter of that method is always self, by convention. This helps Python know that the method belongs to an instance of the class.

* Combining Classes

```python
class Owner:
    def __init__(self, name, address, contact_number):
        self.name = name
        self.address = address
        self.contact_number = contact_number

class Dog:
    def __init__(self, name, breed, owner: Owner):
        self.name = name
        self.breed = breed
        self.owner = owner

    def bark(self):
        print("Woof Woof")

    def get_name(self):
        return self.name
    
    def get_owner_contact(self):
        return self.owner.contact_number

owner = Owner("Danny", "122 Springfield Way, Uk", "8888 888 888")
scottyDog = Dog("Bruce", "Biggs", owner)

scottyDog.bark()
print(f"Dog Name : {scottyDog.get_name()}")
print(f"Get owner's contect number : {scottyDog.get_owner_contact()}")


```

### Access patterns

* How to access ( get/set ) object variable
    * Traditional way is using getter setter methods
    * Using properties

<br>

* Attribute and method accessing pattern
    * Public : Defined as `attribute_name`, accessible from outside, general purpose use.
    * Protected : Defined as `_attribute_name`, accessible but strongly discouraged, internal or subclass use.
    * Private : Defined as `__attribute_name`, not accessible via name mangling.

<br>

* Static 
    * Attribute(Class variable) : To track class level values, like object count, average of a object's attribute
        * Tracking global state
        * Defining class level constants, for e.g. : Minimum balance in a account cannot be less than MINIMUM_BALANCE static attribute
    * Methods : It belongs to the class only.
        * Utiliy functions that the class may possess



```python
# Accessing using @property and setting using @<attribute_name>.setter

class User:

    # Static attribute
    user_count = 0

    @staticmethod
    def concat_values():
        return User.user_count

    def __init__(self, username, email, is_admin = False):
        self.username = username
        self.email = email # this triggers the setter property
        self.__is_admin = is_admin
        User.user_count += 1

    @property
    def email(self):
        if self.__is_admin:
            return self._email # This means it it referencing to the privatised object variable 
        else:
            print("Cannot return email as not admin")

    @email.setter
    def email(self, new_email):
        if "@" in new_email:
            self._email = new_email
        else:
            raise ValueError("Invalid email, missing @")



user1 = User("dantheman", "dan@gmail.com", True)
print(user1.email)

try:
    user1.email = "dayyn@gmail.com"
except ValueError as e:
    print(f"Error: {e}")

print(user1.email)

# Accessing static attribute
print(user1.user_count)

user1.concat_values()
```
